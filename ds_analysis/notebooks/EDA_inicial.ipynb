## 3.5 Distribución de montos

````python
plt.figure(figsize=(10, 6))
plt.hist(df["Amount"], bins=30, color='skyblue', edgecolor='black')
plt.title('Distribución de Montos de Transacción')
plt.xlabel('Monto')
plt.ylabel('Frecuencia')
plt.grid(axis='y', alpha=0.75)
plt.show()
````

### Análisis de la distribución de montos:

*   **¿La mayoría de compras son pequeñas?**

*   **¿Existen valores extremos?**
    *   El histograma también sugiere la presencia de algunos valores atípicos o extremos, con montos de transacción significativamente más altos que la media, lo que se refleja en la cola derecha de la distribución.

# 6. Normalización

Esta sección describe cómo se podrían transformar las columnas existentes para alinearse con un formato estandarizado según las necesidades del contrato o de análisis futuros. Aquí se proponen ejemplos de normalización para algunas columnas clave:

---

### `PaymentMethod`

Los valores actuales (`Debit Card`, `Net Banking`, `Credit Card`, `UPI`, `Cash`) podrían normalizarse a un conjunto más simple y genérico para el contrato de la API:

```
PaymentMethod
---------------
Debit Card   --> DEBITO
Net Banking  --> TRANSFERENCIA
Credit Card  --> CREDITO
UPI          --> TRANSFERENCIA
Cash         --> EFECTIVO
```

---

### `TransactionType`

Los valores actuales (`Debit`, `Credit`) son ya bastante estándar, pero podrían mapearse explícitamente a términos como `EGRESO` e `INGRESO` para mayor claridad en el contexto financiero.

```
TransactionType
---------------
Debit   --> EGRESO
Credit  --> INGRESO
```

---

### `Category`

La columna `Category` ya está disponible y contiene valores útiles. La normalización aquí podría implicar la creación de un catálogo de categorías estandarizado si el contrato define un conjunto específico de categorías que difiere de las que ya existen en el dataset. Por ejemplo:

```
Category
------------------------------------------------
Bills            --> FACTURAS_Y_SERVICIOS
Food             --> ALIMENTACION
Shopping         --> COMPRAS_GENERALES
Online Shopping  --> COMPRAS_ONLINE
Travel           --> VIAJES
...              --> ... (mapear a categorías estandarizadas)
```

# 7. Observaciones

Esta sección destaca puntos clave del dataset que pueden ser de interés para los equipos involucrados, ya sea para el desarrollo de modelos, la mejora de funcionalidades o la identificación de oportunidades.

### Observación 1: Utilización de la variable `Category`

El dataset ya contiene una variable `Category` bien definida. Esto es una ventaja significativa, ya que podría utilizarse directamente como etiqueta (target) para entrenar o validar un clasificador de categorías, evitando la necesidad de inferir inicialmente la categoría únicamente desde la descripción de la transacción. Esto acelera el desarrollo de modelos de categorización y mejora la precisión inicial.

### Observación 2: Variables no contempladas en el contrato actual pero potencialmente útiles

El contrato actual podría no contemplar explícitamente información como `MerchantType`, `DeviceUsed`, `Weekday`, `TimeOfDay`, y `Location`. Sin embargo, estas variables contienen información valiosa que podría ser utilizada para:

*   **Recomendaciones financieras:** Por ejemplo, recomendar promociones de comercios (`MerchantType`) o sugerencias basadas en hábitos de gasto por `Location` o `TimeOfDay`.
*   **Modelos predictivos avanzados:** Predecir patrones de gasto en ciertos `Weekday` o identificar comportamientos de riesgo basados en el `DeviceUsed` o `Location` de la transacción.

Estas variables podrían enriquecer futuras funcionalidades o desarrollar modelos predictivos más sofisticados.

### Observación 3: Variables del contrato no presentes en el dataset

El contrato incorpora variables financieras importantes como `credito_total`, `ingreso_mensual`, `frecuencia_ahorro`, o `nivel_endeudamiento`. Estas no están presentes en el dataset actual, que se enfoca en transacciones individuales. Por lo tanto, para cumplir completamente con el contrato, será necesario:

*   Integrar estas variables mediante un *endpoint* de API adicional.
*   Enriquecer el conjunto de datos con información del perfil financiero del usuario obtenida de otras fuentes.
